During the events, the attendees would have a four minute "first date" with every other participant of the opposite sex. At the end of their four minutes, participants were asked if they would like to see their date again. They were also asked to rate their date on six attributes: Attractiveness, Sincerity, Intelligence, Fun, Ambition, and Shared Interests.

```python
def ident_missing_values(
    df: DataFrame,
    output: Optional[Union[str, list]] = None
) -> Union[dict, Series, int, tuple]:
    """
    Analyse les valeurs manquantes par colonne.
    
    output=None        -> dict complet (toutes les stats), comportement par défaut
    output="miss_0"     -> une seule valeur
    output=["miss_0", "pct_missing"] -> tuple des valeurs demandées, dans l'ordre
    """
    if not isinstance(df, DataFrame):
        raise TypeError("df doit être un DataFrame")
    if df.empty:
        raise ValueError("df est vide")

    pct_missing = (df.isnull().sum() / len(df) * 100).sort_values(ascending=False)

    stats = {
        "pct_missing": pct_missing,
        "miss_0": int((pct_missing == 0).sum()),
        "miss_less5": int(((pct_missing > 0) & (pct_missing < 5)).sum()),
        "miss_less15": int(((pct_missing >= 5) & (pct_missing < 15)).sum()),
        "miss_less40": int(((pct_missing >= 15) & (pct_missing < 40)).sum()),
        "miss_above40": int((pct_missing >= 40).sum()),
    }

    
    if output is None:
        return stats
    if isinstance(output, str):
        if output not in stats:
            raise KeyError(f"clé inconnue : {output}. Choix possibles : {list(stats)}")
        return stats[output]
    # liste de clés -> tuple ordonné
    missing_keys = [k for k in output if k not in stats]
    if missing_keys:
        raise KeyError(f"clés inconnues : {missing_keys}. Choix possibles : {list(stats)}")
    return tuple(stats[k] for k in output)




print(f"Column_name  %_missing" )

stats = ident_missing_values(df)

pct_missing = stats["pct_missing"]
miss_0 = stats["miss_0"]
miss_less5 = stats["miss_less5"]
miss_less15 = stats["miss_less15"]
miss_less40 = stats["miss_less40"]
miss_above40 =  stats["miss_above40"]               

print(f"colonnes à 0% de valeurs manquantes : {stats['miss_0']} : TOP")
print(f"colonnes < 5% manquantes : {stats['miss_less5']} : à garder, imputation mean/mode")
print(f"colonnes 5-15% manquantes : {stats['miss_less15']} : imputation à revoir (KNN, MICE, plutôt modèle prédictif que mean/median)")
print(f"colonnes 15-40% manquantes : {stats['miss_less40']} : à supprimer sauf pattern exploitable (penser MCAR/MAR/MNAR)")
print(f"colonnes > 40% manquantes : {stats['miss_above40']} : à supprimer, sauf variable jugée précieuse pour l'analyse")


```

```python
# fill missing values with average note : 5/10 if 1-10 scale / or 
# or fill with mean of the column for each wave, or fill with mean of the column for each gender
mask_10 = df['wave'].isin(ten_scale_waves)

# 1) waves à échelle 0-10 : NaN -> 5
df.loc[mask_10, sc_rat_cols] = df.loc[mask_10, sc_rat_cols].fillna(5)
df.loc[mask_10, sc_partner_cols] = df.loc[mask_10, sc_partner_cols].fillna(5)

# 2) autres waves : NaN -> médiane de la colonne (calculée sur ces mêmes waves)
medians = df.loc[~mask_10, sc_rat_cols].median()
df.loc[~mask_10, sc_rat_cols] = df.loc[~mask_10, sc_rat_cols].fillna(medians)

medians = df.loc[~mask_10, sc_partner_cols].median()
df.loc[~mask_10, sc_partner_cols] = df.loc[~mask_10, sc_partner_cols].fillna(medians)   

# 3) rescale les lignes des autres waves pour que la somme des sc_rat_cols = 100
row_sums = df.loc[~mask_10, sc_rat_cols].sum(axis=1)
df.loc[~mask_10, sc_rat_cols] = df.loc[~mask_10, sc_rat_cols].div(row_sums, axis=0) * 100

row_sums = df.loc[~mask_10, sc_partner_cols].sum(axis=1)
df.loc[~mask_10, sc_partner_cols] = df.loc[~mask_10, sc_partner_cols].div(row_sums, axis=0) * 100


# df.loc[df.wave==8,sc_rat_cols].isnull()=
```

#### biserial correlations
Let's try a 2nd method : 
compute biserial correlations between order_rate (numerical variable) and decision (boolean variable)
 

```python
# method 2 : when computing biserial correlations between order_rate (numerical variable) and decision (boolean variable)
from scipy.stats import pointbiserialr
import pandas as pd

def corr_num_bool(df, num_col, bool_col):
    """
    Corrélation point-bisériale entre une variable numérique
    et une variable booléenne/binaire (0/1 ou True/False).
    """
    sub = df[[num_col, bool_col]].dropna()
    # S'assurer que le booléen est bien en 0/1
    bool_vals = sub[bool_col].astype(int)
    r, p = pointbiserialr(bool_vals, sub[num_col])
    return {"r": r, "p_value": p, "n": len(sub)}

result = corr_num_bool(df, "order_rate", "dec")
print(result)
print(f"")
result2 = corr_num_bool(df, "order_rate", "dec_o")
print(result2)
result3 = corr_num_bool(df, "order_rate", "match")
print(result3)
```

Results of the test show that correlation is very low, or null 
No clear link can be seen on the global data

p ≈ 0.18 : non significatif (> 0.05) → on ne peut pas rejeter l'hypothèse nulle d'absence de corrélation
n = 8378 : échantillon large, donc le test a beaucoup de puissance statistique — s'il y avait un effet réel même faible, il aurait probablement été détecté
Conclusion

Avec un tel échantillon, la non-significativité est parlante : la variable numérique testée et la variable booléenne sont statistiquement indépendantes (ou l'effet est tellement faible qu'il est négligeable en pratique, même s'il existait).
